# 1. Configurando a Sessão do Apache Iceberg

```python
from pyspark.sql import SparkSession

# Inicializando a sessão Spark com as configurações do Iceberg
spark = SparkSession.builder \
    .appName("IcebergLocalDevelopment") \
    .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.4_2.12:1.4.1") \
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
    .config("spark.sql.catalog.local", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.local.type", "hadoop") \
    .config("spark.sql.catalog.local.warehouse", "spark-warehouse/iceberg") \
    .getOrCreate()

spark.conf.set("spark.sql.shuffle.partitions", "1")
print("Sessão Spark com Iceberg iniciada com sucesso!")
```

# 2. Criando a Tabela Iceberg (DDL)

```python
spark.sql("""
    CREATE TABLE IF NOT EXISTS local.db.estudantes (
        ID_ALUNO INT,
        NOME_ALUNO STRING,
        GRADE STRING,
        CPF STRING
    ) USING iceberg
""")
print("Tabela 'estudantes' garantida no catálogo Iceberg.")
```

# 3. Inserindo os Dados Iniciais (DML - Insert)

```python
spark.sql("""
    INSERT INTO local.db.estudantes VALUES
    (1, 'Luis', 'FUNDAMENTAL', '564.654.654-21'),
    (2, 'Jonh', 'POS-GRADUACAO', '321.497.312-21'),
    (3, 'Hanz', 'MEDIO', '978.348.914-45'),
    (4, 'Ana', 'POS-GRADUACAO', '321.497.312-21')
""")

print("Dados inseridos:")
spark.sql("SELECT * FROM local.db.estudantes").show()
```

# 4. Modificando os Dados (DML - Update e Delete)

```python
spark.sql("DELETE FROM local.db.estudantes WHERE ID_ALUNO = 1")
spark.sql("UPDATE local.db.estudantes SET GRADE = 'FORMADA' WHERE ID_ALUNO = 4")

print("Dados após UPDATE e DELETE:")
spark.sql("SELECT * FROM local.db.estudantes").show()
```
